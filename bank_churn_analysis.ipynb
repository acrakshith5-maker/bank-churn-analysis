```python
import matplotlib.pyplot as plt

df.groupby('churn')['active_member'].mean().plot(kind='bar', title='Active Membership Rate by Churn Status')
plt.xlabel('Churn (0 = Stayed, 1 = Left)')
plt.ylabel('Proportion Active')
plt.show()
```

```python
df.groupby('churn')['balance'].mean().plot(kind='bar', title='Average Balance by Churn Status', color=['steelblue','indianred'])
plt.xlabel('Churn (0 = Stayed, 1 = Left)')
plt.ylabel('Average Balance (€)')
plt.show()
```
